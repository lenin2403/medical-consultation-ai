```python
# Case Study 2 - Data Lab Environment Check

import sys
import platform
import shutil
import subprocess
from pathlib import Path

print("DATA LAB ENVIRONMENT CHECK")
print("=" * 60)

# Python / OS
print("Python version:", sys.version)
print("Operating system:", platform.platform())

# PyTorch and GPU
try:
    import torch

    print("\nPyTorch version:", torch.__version__)
    print("CUDA available:", torch.cuda.is_available())

    if torch.cuda.is_available():
        print("CUDA version:", torch.version.cuda)
        print("GPU count:", torch.cuda.device_count())

        for i in range(torch.cuda.device_count()):
            properties = torch.cuda.get_device_properties(i)

            print(f"\nGPU {i}: {torch.cuda.get_device_name(i)}")
            print(
                "GPU memory:",
                round(properties.total_memory / 1024**3, 2),
                "GB"
            )

except Exception as error:
    print("\nPyTorch/GPU check error:", error)

# NVIDIA system information
print("\n" + "=" * 60)
print("NVIDIA-SMI")
print("=" * 60)

nvidia = subprocess.run(
    ["nvidia-smi"],
    capture_output=True,
    text=True
)

print(nvidia.stdout)

if nvidia.stderr:
    print("NVIDIA error:")
    print(nvidia.stderr)

# FFmpeg
print("=" * 60)
print("FFMPEG")
print("=" * 60)

ffmpeg_path = shutil.which("ffmpeg")

print("FFmpeg available:", ffmpeg_path is not None)

if ffmpeg_path:
    ffmpeg = subprocess.run(
        ["ffmpeg", "-version"],
        capture_output=True,
        text=True
    )

    print(ffmpeg.stdout.splitlines()[0])

# RAM
print("\n" + "=" * 60)
print("SYSTEM MEMORY")
print("=" * 60)

try:
    with open("/proc/meminfo", "r") as file:
        first_line = file.readline()

    ram_kb = int(first_line.split()[1])

    print(
        "Total RAM:",
        round(ram_kb / 1024**2, 2),
        "GB"
    )

except Exception as error:
    print("RAM check error:", error)

# Storage
print("\n" + "=" * 60)
print("STORAGE")
print("=" * 60)

disk = shutil.disk_usage("/")

print(
    "Total storage:",
    round(disk.total / 1024**3, 2),
    "GB"
)

print(
    "Free storage:",
    round(disk.free / 1024**3, 2),
    "GB"
)

print(
    "Used storage:",
    round(disk.used / 1024**3, 2),
    "GB"
)

# Current working directory
print("\n" + "=" * 60)
print("CURRENT LOCATION")
print("=" * 60)

print("Current directory:", Path.cwd())

print("\nTop-level folders:")

for item in sorted(Path.cwd().iterdir()):
    print("-", item.name)
```
